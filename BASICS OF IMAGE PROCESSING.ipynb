```python
import cv2 as cv
import matplotlib.pyplot as plt
import numpy as np
```

```python
lena=cv.imread('Lenna_(test_image).png')
```

```python
lena
```

```python
plt.imshow(lena)
```

```python
image = lena

half = cv.resize(image, (0, 0), fx = 0.4, fy = 0.4)
bigger = cv.resize(image, (1050, 1610))

stretch_near = cv.resize(image, (300, 300), 
               interpolation = cv.INTER_LINEAR)


Titles =["Original", "Half", "Bigger", "Interpolation Nearest"]
images =[image, half, bigger, stretch_near]
count = 4

for i in range(count):
    plt.subplot(2, 2, i + 1)
    plt.title(Titles[i])
    plt.imshow(images[i])

plt.show()

```

### Rgb to Grey

```python
lena_grey=cv.cvtColor(lena,cv.COLOR_RGB2GRAY)
plt.title("Grey")
plt.imshow(lena_grey,cmap='grey')

plt.show()
```

##### Normal avg

```python
img=lena
(row, col) = img.shape[0:2]


for i in range(row):
    for j in range(col):
        img[i, j] = sum(img[i, j]) * 0.33

plt.imshow(img)
```

```python
(row, col) = img.shape[0:2]

for i in range(row):
    for j in range(col):
        r, g, b = img[i, j]
        
        gray_value = int(b*0.299 + g*0.587 + r*0.114)
        
        img[i, j] = [gray_value, gray_value, gray_value]

plt.imshow(cv.cvtColor(img, cv.COLOR_BGR2RGB))  
plt.axis('off')  
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

img =lena
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

equalized_img = cv2.equalizeHist(gray_img)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(gray_img, cmap='gray')
plt.title('Original Grayscale Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(equalized_img, cmap='gray')
plt.title('Equalized Grayscale Image')
plt.axis('off')

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.hist(gray_img.ravel(), bins=256, color='gray', alpha=0.7)
plt.title('Histogram of Original Grayscale Image')

plt.subplot(1, 2, 2)
plt.hist(equalized_img.ravel(), bins=256, color='gray', alpha=0.7)
plt.title('Histogram of Equalized Grayscale Image')

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

img =lena

b, g, r = cv2.split(img)

b_eq = cv2.equalizeHist(b)
g_eq = cv2.equalizeHist(g)
r_eq = cv2.equalizeHist(r)

equalized_img = cv2.merge([b_eq, g_eq, r_eq])

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))  
plt.title('Original Color Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(equalized_img, cv2.COLOR_BGR2RGB))  
plt.title('Equalized Color Image')
plt.axis('off')


plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.hist(b.ravel(), bins=256, color='blue', alpha=0.7)
plt.title('Histogram of Original Blue Channel')

plt.subplot(2, 2, 2)
plt.hist(g.ravel(), bins=256, color='green', alpha=0.7)
plt.title('Histogram of Original Green Channel')

plt.subplot(2, 2, 3)
plt.hist(r.ravel(), bins=256, color='red', alpha=0.7)
plt.title('Histogram of Original Red Channel')
plt.subplot(2, 2, 4)
plt.hist(equalized_img.ravel(), bins=256, color='gray', alpha=0.7)
plt.title('Histogram of Equalized Image (All Channels)')

plt.show()
```
